```python
import colorama
from colorama import Fore, Style
from textblob import TextBlob
```

```python
#Initialize colorama for colored output
colorama.init()
```

```python
#Emojis for the start of the program
print(f"{Fore.CYAN} Welcome to Sentiment Spy! {Style.RESET_ALL}")
```

```python
user_name = input(f"{Fore.MAGENTA} Please enter your name: {Style.RESET_ALL}").strip()
if not user_name:
    user_name = "Mystery Agent"

```

```python
#Store conversation as a list of tuples (text, polarity, sentiment_type)
conversation_history = []

print(f"\n {Fore.CYAN} Hello, Agent {user_name}!")
print("Type a sentence and I will analyze it using TextBlob")
print(f"Type {Fore.YELLOW} reset {Fore.CYAN},{Fore.YELLOW}history{Fore.CYAN},", f"or {Fore.YELLOW}exit{Fore.CYAN} to quit. {Style.RESET_ALL} \n")

while True:
    user_input = input(f"{Fore.GREEN} >> {Style.RESET_ALL}").strip()

    if not user_input:
        print(f"{Fore.RED} Please enter text or a valid command. {Style.RESET_ALL}")
        continue

    #Check for commands
    if user_input.lower() == "exit":
        print(f"\n {Fore.BLUE} Exiting Sentiment Spy. Farewell, Agent {user_name}! {Style.RESET_ALL}")
        break

    elif user_input.lower() == "reset":
        conversation_history.clear()
        print(f"{Fore.CYAN} All conversation history cleared! {Style.RESET_ALL}")
        continue

    elif user_input.lower() == "history":
        if not conversation_history:
            print(f"{Fore.YELLOW} No conversation history yet. {Style.RESET_ALL}")
        else:
            print(f"{Fore.CYAN} Conversation history: {Style.RESET_ALL}")

```
